## Q7. Sum of weights

1. Select all the cars from Asia.
2. Select only columns `vehicle_weight` and `model_year`.
3. Select the first 7 values.
4. Get the underlying NumPy array (`X`).
5. Compute matrix-matrix multiplication between the transpose of `X` and `X` (`XTX`).
6. Invert `XTX`.
7. Create an array `y` with values `[1100, 1300, 800, 900, 1000, 1100, 1200]`.
8. Multiply the inverse of `XTX` with the transpose of `X`, and then multiply the result by `y` to get `w`.
9. What's the sum of all the elements of `w`?

```python
# Your code here:
X = df[df["origin"] == "Asia"][["vehicle_weight", "model_year"]].head(7).to_numpy()
X
```

```python
XTX = X.T @ X
XTX
```

```python
XTX_inv = np.linalg.inv(XTX)
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
w = XTX_inv @ X.T @ y
w
```

```python
w.sum()
```
